```python
import spacy

# Load spaCy's lightweight English core model (~12MB)
nlp = spacy.load("en_core_web_sm")

# Sample unstructured text input
raw_resume_text = """
Veer Raghuvanshi is an accomplished Software Developer based in Ghaziabad,
bringing a strong academic foundation from SRM to his enterprise engineering work.
Currently, he drives technical innovation at Google, where he has been an integral part of the team since March 2027.
With core expertise in Python, Machine Learning, and Amazon Web Services (AWS), 
Veer specializes in designing intelligent, scalable cloud architectures.
"""

print("Processing text with spaCy...\n")
doc = nlp(raw_resume_text)

print("--- Extracted Named Entities ---")
# Extract entities like Persons, Organizations, Locations, and Dates
for ent in doc.ents:
  print(f"Entity: {ent.text:<20} | Type: {ent.label_:<10} ({spacy.explain(ent.label_)})")

print("\n--- Identified Keywords/Nouns ---")
# Extract key nouns/skills
nouns = [token.text for token in doc if token.pos_ in ["NOUN", "PROPN"]]
print("Key Nouns Found:", list(set(nouns)))
```
