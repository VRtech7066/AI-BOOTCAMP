```python
import numpy as np
players = np.array(['Sachin', 'Virat Kohli', 'Yashasvi', 'Jaspreet', 'Rishab'])
matches = np.array(['IPL 2015', 'IPL 2020', 'IPL 2023'])
np.random.seed(42)
runs = np.random.randint(10, 100, size=(len(players), len(matches)))
print("Players:", players)
print("Matches:", matches)
print("Runs scored by each player:")
print(runs)
```

```python
#Calculate the standard deviation of runs for each player
std_dev_by_player = runs.std(axis=1)
print("Variability in perormance (Standard Deviation):")
for player, std in zip(players, std_dev_by_player):
    print(f"{player}: {std:.2f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define data as a single 2D Numpy array
# Rows = Students (6), Columns = Hobbies (5)
data = np.array([
    [8, 6, 9, 5, 7],
    [7, 9, 6, 8, 5],
    [9, 7, 8, 6, 9],
    [6, 8, 7, 9, 6],
    [8, 5, 9, 7, 8],
    [7, 8, 8, 6, 7]
])

hobbies = ["Running", "Gaming", "Music", "Sports", "Drawing"]
students = [f"Student {i+1}" for i in range(data.shape[0])]

# 2. Numpy Calculations
# axis=0 calculates down the columns (Hobbies)
# axis=1 calculates across the rows (Students)
hobby_means = np.mean(data, axis=0)
hobby_medians = np.median(data, axis=0)
hobby_stds = np.std(data, axis=0)

student_means = np.mean(data, axis=1)

print("COLUMN-WISE STATISTICS (Hobbies)")
print("-" * 50)
for i, hobby in enumerate(hobbies):
    print(f"{hobby:10}: Mean={hobby_means[i]:.2f}, Median={hobby_medians[i]:.2f}, Std={hobby_stds[i]:.2f}")

# 3. Visualization: Grouped Bar Chart for Hobbies
x = np.arange(len(hobbies))
width = 0.25

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Chart 1: Hobby Statistics
ax1.bar(x - width, hobby_means, width, label='Mean', color='#4C72B0')
ax1.bar(x, hobby_medians, width, label='Median', color='#55A868')
ax1.bar(x + width, hobby_stds, width, label='Std Dev', color='#C44E52')

ax1.set_ylabel('Scores')
ax1.set_title('Statistics by Hobby')
ax1.set_xticks(x)
ax1.set_xticklabels(hobbies)
ax1.legend()

# Chart 2: Student Averages
ax2.bar(students, student_means, color='#8172B3')
ax2.set_title('Average Score per Student')
ax2.set_ylabel('Mean Score')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# STEP 1:
time = np.linspace(0, 10, 500)
damping = np.exp(-0.3 * time) 
amplitude = damping * np.sin(2 * np.pi * time)
```

```python
# STEP 2:
fig, ax = plt.subplots(figsize=(8, 4.5), dpi=300)
```

```python
# STEP 3: 
ax.plot(time, amplitude, color='crimson', linewidth=2, label="Sensor A Displacement")
ax.set_title('Damped Harmonic Response of Structure', fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Time (seconds)', fontsize=12)
ax.set_ylabel('Displacement (mm)', fontsize=12)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Generate engineering data using NumPy
time = np.linspace(0, 10, 500)  # 500 points from 0 to 10 seconds
amplitude1 = np.sin(2 * np.pi * time)  # Simple sine wave
amplitude2 = np.exp(-0.5 * time) * np.sin(2 * np.pi * time)  # Damped sine wave

# 2. Create a figure with 2 subplots vertically (2 rows, 1 column)
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 8), dpi=100)

# 3. Plot 1: Simple Sine Wave (Top Subplot)
ax1.plot(
    time,
    amplitude1,
    color="crimson",
    linewidth=2,
    label="Undamped Harmonic Motion",
)
ax1.set_title(
    "Harmonic Response of Structure", fontsize=14, fontweight="bold", pad=12
)
ax1.set_xlabel("Time (seconds)", fontsize=11)
ax1.set_ylabel("Displacement (mm)", fontsize=11)
ax1.grid(True, linestyle="--", alpha=0.6)
ax1.legend(loc="upper right", frameon=True, shadow=True)

# 4. Plot 2: Damped Sine Wave (Bottom Subplot)
ax2.plot(
    time, amplitude2, color="navy", linewidth=2, label="Damped Harmonic Motion"
)
ax2.set_title(
    "Damped Harmonic Response of Structure", fontsize=14, fontweight="bold", pad=12
)
ax2.set_xlabel("Time (seconds)", fontsize=11)
ax2.set_ylabel("Displacement (mm)", fontsize=11)
ax2.grid(True, linestyle="--", alpha=0.6)
ax2.legend(loc="upper right", frameon=True, shadow=True)

# 5. Adjust layout to avoid overlapping labels and display
plt.tight_layout()
# plt.savefig('sensor_data_plot.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

x = np.linspace(0, 2*np.pi, 200)
fig, ax = plt.subplots()
line, = ax.plot(x, np.sin(x), lw=3)
ax.set_ylim(-1.2, 1.2)

def update(phase):
    line.set_ydata(np.sin(x + phase))
    return line,

ani = FuncAnimation(fig, update, frames=np.linspace(0, np.pi/2, 60),
                    interval=40, blit=True)

HTML(ani.to_jshtml())
```
