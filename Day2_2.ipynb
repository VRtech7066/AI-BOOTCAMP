```python
#Cricket Analytical Data 
import pandas as pd
import numpy as np

# Same random data every time
np.random.seed(42)

n = 100

# Player names
players = [f"Player_{i:03d}" for i in range(1, n + 1)]

# Generate cricket data
age = np.random.randint(19, 39, n)
matches = np.random.randint(20, 151, n)

innings = np.maximum(
    matches - np.random.randint(0, 15, n),
    1
)

runs = np.random.randint(300, 6501, n)

balls_faced = np.maximum(
    runs + np.random.randint(100, 1800, n),
    runs + 1
)

fifties = np.random.randint(0, 36, n)
hundreds = np.random.randint(0, 16, n)
wickets = np.random.randint(0, 121, n)
catches = np.random.randint(0, 81, n)

# Calculate Strike Rate
strike_rate = (runs / balls_faced) * 100
strike_rate = np.round(strike_rate, 2)

# Calculate Batting Average
outs = np.random.randint(5, innings + 1)
batting_average = runs / outs
batting_average = np.round(batting_average, 2)

# Performance Score
performance_score = (
    runs * 0.015
    + strike_rate * 0.35
    + batting_average * 0.8
    + hundreds * 2.5
    + fifties * 0.5
    + wickets * 0.12
    + catches * 0.08
)

performance_score = np.round(performance_score, 2)

# Create DataFrame
df = pd.DataFrame({
    "Player": players,
    "Age": age,
    "Matches": matches,
    "Innings": innings,
    "Runs": runs,
    "Balls_Faced": balls_faced,
    "Batting_Average": batting_average,
    "Strike_Rate": strike_rate,
    "Fifties": fifties,
    "Hundreds": hundreds,
    "Wickets": wickets,
    "Catches": catches,
    "Performance_Score": performance_score
})

# Save as CSV
df.to_csv("cricket_player_ml_dataset.csv", index=False)

# Display first 10 rows
print(df.head(10))

print("\nCSV file created successfully!")
```
